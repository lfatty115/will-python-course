```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(df_ne['Time period'], df_ne['North East'], color='royalblue')
plt.title("North East Average House Price Over Time")
plt.xlabel("Date")
plt.ylabel("Average Price (£)")
plt.grid(True)
plt.show()
```

```python
# --- A: Plot price levels ---
plt.figure(figsize=(12,6))
plt.plot(df['Time period'], df['North East'], label='North East')
plt.plot(df['Time period'], df['England'], label='England')
plt.plot(df['Time period'], df['United Kingdom'], label='UK')
plt.legend()
plt.grid(True)
plt.show()

# --- B: Prepare YoY and rolling metrics ---
df['NE_yoy'] = df['North East'].pct_change(12) * 100
df['England_yoy'] = df['England'].pct_change(12) * 100
df['UK_yoy'] = df['United Kingdom'].pct_change(12) * 100

df['NE_roll'] = df['North East'].rolling(6).mean()


```

```python
plt.figure(figsize=(12,6))
plt.plot(df['Time period'], df['NE_yoy'], label='North East YoY')
plt.plot(df['Time period'], df['England_yoy'], label='England YoY')
plt.plot(df['Time period'], df['UK_yoy'], label='UK YoY')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df['Time period'], df['NE_roll'], label='North East (6‑month rolling)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np

# Fit a simple linear trend
x = np.arange(len(df))
y = df['North East'].values

coef = np.polyfit(x, y, 1)
trend = np.poly1d(coef)

# Predict next 12 months
future_x = np.arange(len(df), len(df) + 12)
future_pred = trend(future_x)
```
